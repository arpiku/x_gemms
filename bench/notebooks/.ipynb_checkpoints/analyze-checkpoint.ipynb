# x_gemms Benchmark Analysis

Interactive analysis notebook for GEMM benchmark results.

```python
import os
import sys
sys.path.insert(0, '..')

from bench.analyzer import BenchmarkAnalyzer, RESULTS_DIR
import plotly.io as pio
pio.renderers.default = 'notebook'

print(f"Results directory: {RESULTS_DIR}")
```

## Load Data

```python
# Load latest results or specify files
csv_files = os.environ.get('X_GEMMS_CSV_FILES', '')
if csv_files:
    files = csv_files.split(',')
    analyzer = BenchmarkAnalyzer.from_multiple([f'../../{f}' for f in files])
    print(f"Loaded files: {files}")
else:
    analyzer = BenchmarkAnalyzer.from_latest()
    print(f"Loaded latest results")

print(f"Metadata: {analyzer.metadata}")
print(f"\nLoaded {len(analyzer.results)} results")
print(f"Algorithms: {analyzer.get_algorithms()}")
print(f"Sizes: {analyzer.get_sizes()}")
```

## Summary Statistics

```python
analyzer.get_summary()
```

```python
analyzer.get_best_performers()
```

## GFLOPS Performance

```python
fig = analyzer.plot_gflops()
fig.show()
```

## Memory Bandwidth

```python
fig = analyzer.plot_bandwidth()
fig.show()
```

## Algorithm Comparison

```python
# Compare specific algorithms - modify this list
algorithms_to_compare = ['cublas', 'strassen_cublas', 'winograd_cublas']
fig = analyzer.plot_comparison(algorithms_to_compare)
fig.show()
```

## Speedup Analysis

```python
# Speedup relative to baseline - modify baseline algorithm
baseline = 'cublas'
fig = analyzer.plot_speedup(baseline)
fig.show()
```

## Heatmap View

```python
fig = analyzer.plot_heatmap(metric='gflops')
fig.show()
```

## Filtered Analysis

```python
# Filter to specific size range or algorithms
filtered = analyzer.filter(sizes=(256, 4096), dtypes=['fp32'])
print(f"Filtered to {len(filtered.results)} results")
fig = filtered.plot_gflops()
fig.show()
```
